```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

#### Plot Vocabulary:
-   figure (fig) - container, window to put your plots on
-   axes (ax) - area within the fig where data is plotted
-   axis - x, y axis of a plot or chart
-   line - data itself, cant be fully customized
-   Legend - small container with labels/descriptors of elements
-   labels - names/labels for the x and y axis
-   title - title for the chart/plot
-   markers - symbols that represent data points
-   grid - backgrod (attribute that can be edited)
-   ticks (tic) - marks along the x,y axis to represent values

```python
# Creates a simple line plot
# define the data  on the x and y axes

x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# create a line plot with a dashdot linestyle
plt.plot(x, y, linestyle='-.', marker='.', markersize='10', color ='brown')
plt.show()
```

```python
#Create subplots
fig, ax = plt.subplots()

ax.plot(x, y, linestyle='-.', marker='s', color = 'red')
plt.show()
```

```python
plt.plot(x, y, linestyle=':')
plt.show()
```

```python
x = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])

# Create a dashed line, with diamond markers
plt.plot(y, linestyle='--', marker='v')
plt.show()
```

```python
x = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])

# Using the opposite approach, linestyle of dashdot, and square markers
fig, ax = plt.subplots()

ax.plot(y, linestyle='-.', marker='D')
plt.show()
```

```python
# Create a lineplot, custom cuolor, dotted linestyle, diamond markers
# create line plot dotted line,marker diamond,color
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
plt.plot(
    x, y,
    linestyle=':',
    marker='D',
    color=(0.5, 0.5, 0.8)
)

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.show()
```

```python
# Create a lineplot, custom cuolor, dotted linestyle, diamond markers
plt.plot(
    x, y,
    linestyle=':',
    marker='D',
    color='#FF5733'
)
plt.title('Sports Watch Data', color='blue')
plt.xlabel('Average Pulse', color='green')
plt.ylabel('Calorie Burnage', color='red')
plt.show()
```

```python
years=[1970,1980,1985,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
pops=[3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.3,7.4,7.5]
```
